```python
num=[]
for i in range(10):
 n=int(input("enter value:"))
 num.append(n)
print("\nlist:",num)

#a)count number of occurences

for i in range(10):
 if num.count(num[i])>1:
     print(num[i],"->",end="")

     for j in range(10):
         if num[j]==num[i]:
             print(j,end="")

print()

#b)count even and odd vlues
even=0
odd=0

for n in num:
 if n%2==0:
    even+=1
 else:
    odd+=1
print("\nEven:",even)
print("\nOdd:",odd)

#c)count positive and negative
pos=0
neg=0

for n in num:
 if n>0:
    pos+=1
 elif n<0:
    neg+=1
print("\nPositive:",pos)
print("\nNegative:",neg)
```
