```python
import math

def input_sides():
    a=float(input("enter side a"))
    b=float(input("enter side b"))
    c=float(input("enter side c"))
    return a,b,c

def check_tri(a,b,c):
    if a*a + b*b ==c*c or c*c + b*b ==a*a or a*a + c*c ==b*b:
        print("It is a right angle triangle")
    else:
        print("Not a right angle triangle")

a,b,c=input_sides()
check_tri(a,b,c)
```
