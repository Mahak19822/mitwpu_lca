```python
student_list=["mahak","ayush","soumya","prachi"]
print("\nstudent list:")
print(student_list)

student_tuple=("mahak","ayush","soumya","prachi")
print("\nstudent tuple:")
print(student_tuple)

student_dict={
    1:"Mahak",
    2:"ayush",
    3:"soumya",
    4:"prachi",
}
print("\nstudent dictionary:")
print(student_dict)

#addition
student_dict[5]="neha"
print(student_dict)

#updation
student_dict[2]="rohan"
print(student_dict)

#deletion
del student_dict[4]
print(student_dict)
```
